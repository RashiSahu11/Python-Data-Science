# MATPLOTLIB

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
df
```

```python
#histogram -distribution of sepal length
plt.hist(df['sepal_length'],color='purple',edgecolor='black')
plt.title("Distribution of sepal length")
plt.xlabel("sepal length")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.plot(df['petal_length'],color='orange')
plt.title("petal_length ")
plt.xlabel("sample index")
plt.ylabel("petal length")
plt.show()
```

```python
plt.plot(df['petal_length'][:50],color='red')
plt.title("petal_length of first 50 samples ")
plt.xlabel("sample index")
plt.ylabel("petal length")
plt.show()
```

```python
mean_value=df.groupby("species").mean()
mean_value
```

```python
mean_value.plot(kind='bar',figsize=(5,5))
plt.title("Average Feature values per species")
plt.xlabel("species")
plt.ylabel("average value")
plt.xticks(rotation=20)
plt.show()
```

```python
fig,ax=plt.subplots(figsize=(8,3))
#setting color figure and axes
fig.set_facecolor("skyblue")
ax.set_facecolor("lightyellow")
#setting the title and label
ax.set_title("Customized line plot",color='red',fontsize=18,
                fontfamily='serif',fontweight='bold',loc='right')
ax.set_xlabel("X-axis",color='green',fontsize=15,fontfamily='serif')
ax.set_ylabel("Y-axis",color='green',fontsize=15,fontfamily='serif')
#data plotting
x=[1,2,3,4,5,6,7,8,9]
y=[3,6,2,8,3,9,5,1,4] 
ax.plot(x,y,
        color='red',
        linewidth=1,
        linestyle='--',
        marker='o',
        markersize=6,
        markerfacecolor='yellow',
        markeredgecolor='black',
        markeredgewidth=0.4
)

#adding text
ax.text(3,7,"sample text",fontsize=12,color='blue',fontfamily='serif',fontweight='bold')

#setting partition
ax.axhline(y=5,color='black',alpha=0.4)
ax.axvline(x=5,color='black')
```

```python
fig,ax=plt.subplots(figsize=(8,3))
#setting color figure and axes
fig.set_facecolor("blue")
ax.set_facecolor("yellow")
#setting the title and label
ax.set_title("Customized line plot",color='red',fontsize=18,
                fontfamily='serif',fontweight='bold',loc='right')
ax.set_xlabel("X-axis",color='green',fontsize=15,fontfamily='serif')
ax.set_ylabel("Y-axis",color='green',fontsize=15,fontfamily='serif')
#data plotting
x=[1,2,3,4,5,6,7,8,9]
y=[2,4,6,8,2,4,6,8,2] 
ax.plot(x,y,
        color='purple',
        linewidth=1,
        linestyle=':',
        marker='d',
        markersize=6,
        markerfacecolor='orange',
        markeredgecolor='black',
        markeredgewidth=0.4
        )

#adding text
ax.text(3,7,"sample text",fontsize=12,color='blue',fontfamily='serif',fontweight='bold')

#setting partition
ax.axhline(y=5,color='black',alpha=0.4)
ax.axvline(x=5,color='black')
```

```python
titanic=sns.load_dataset('titanic')
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(titanic['fare'], titanic['age'], alpha=0.5, c=titanic['age'], cmap='viridis')
plt.title('Fare vs Age')
plt.xlabel('Fare')
plt.ylabel('Age')
plt.colorbar(label='Age')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
