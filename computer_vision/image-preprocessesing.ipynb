```python
w,h=img.size
print(w,h)
```

```python
small_img=img.resize((w//2,h//2))
small_img
```

```python

img.resize((w*2,h*2),Image.Resampling.LANCZOS)
```

```python

cropped_img=small_img.crop((190,150,280,280))
cropped_img
```

```python
small_img.filter(ImageFilter.BoxBlur(2))
```

```python
small_img.filter(ImageFilter.BLUR)
```

```python
small_img.filter(ImageFilter.DETAIL)
```

```python
small_img.filter(ImageFilter.CONTOUR)
```

```python
small_img.filter(ImageFilter.EMBOSS)
```

```python
small_img.rotate(180)
```

```python
small_img.transpose(Image.FLIP_LEFT_RIGHT)
```

```python
small_img.transpose(Image.FLIP_TOP_BOTTOM)
```

```python
from PIL import ImageEnhance
```

```python
enhancer=ImageEnhance.Brightness(small_img)
```

```python
enhancer.enhance(1.9)
```

```python
contrast=ImageEnhance.Contrast(small_img)
```

```python
contrast.enhance(1.9)
```

```python
from PIL import ImageFont,ImageDraw
```

```python
font=ImageFont.truetype('font.ttf',40)
draw=ImageDraw.draw(small_img)
draw.text((100,50),'Hello world',font=font,fill='black')
```
