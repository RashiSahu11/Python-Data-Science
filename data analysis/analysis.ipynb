```python
import seaborn as sns
sns.get_dataset_names()
```

- When a column has <40% missing values of the data, then we simply remove that  column
- When a column has >40% missing values of data, then we handle those missing values means we replace those missing values with any number
    - there are teo type of columns in any dataset - CATAGORICAL & NUMERICAL
    - When the column is numerical, we always repalce the null values with the column mean(average)
    - When the column is categorical, we always replace the null values with most frequent(mode) values

```python
titanic=sns.load_dataset('titanic')
titanic.isnull().sum()
```

```python
titanic=titanic.dropna(subset=['embarked'])# dropna deletes the null rows automatically
titanic.isnull().sum()
```

```python
# calculate the average age of passenger in the titanic dataset
avg_age=round(titanic['age'].mean()) #round is used to round the value, 
print(avg_age)                               #mean is used to calculate the average
```

```python
#replace null values with avg_age
titanic['age']=titanic['age'].fillna(avg_age)# handle nulll values 
titanic.isnull().sum()
```

```python
#checking duplicates
duplicates=titanic[titanic.duplicated()]
print(duplicates)
```

```python
#dropping duplicates
titanic.drop_duplicates()
```

```python
titanic['age'].unique()
```

```python
#frequence of values in a categorial column
titanic['embark_town'].value_counts()
```

```python
titanic['embark_town'].value_counts().idxmax()
```

```python
titanic['embark_town'].value_counts().idxmin()
```

```python
titanic.sort_index()# used to sort id which we set as index before
```

DATA MANIPULATION

```python
#Adding a new column
titanic = titanic.assign(Discount=titanic['fare']*0.10)
titanic
```

```python
titanic['fare'] = titanic['fare'].apply(lambda x:x*1.05)
titanic
```

```python
titanic['embarked']=titanic['embarked'].map({
    "S":"Southampton",
    "C":"Cherbourg",
    "Q":"Queenstown"
})
titanic
```

```python
titanic.groupby("embarked")['fare'].mean()
```

```python
titanic.groupby("sex")['fare'].max()
```

```python
#data filtering
titanic[titanic['survived']==0]
```

```python
titanic[titanic['sex']=='female']
```

```python
#sorting values
titanic.sort_values("fare",ascending=False)#false arrange data in desending to ascending
                                           #true arrange the data in ascending to descending
```
