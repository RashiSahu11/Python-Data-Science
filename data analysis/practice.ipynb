```python
def print_filled_inverted_triangle(height):
    for i in range(height, 0, -1):
        # Create outer spacing to center the triangle
        spaces = " " * (height - i)
        # Create the filled row of stars
        stars = "*" * (2 * i - 1)
        
        print(spaces + stars)

# Run the function with a height of 5
print_filled_inverted_triangle(5)
```

```python
def get_words_by_criteria(words, criteria=None):

    vowels = "aeiou"

    if criteria == "continuous":
        return [word for word in words if word.lower().endswith("ing")]

    elif criteria == "vowel_rich":
        result = []
        for word in words:
            count = 0
            for ch in word.lower():
                if ch in vowels:
                    count += 1
            if count > 5:
                result.append(word)
        return result

    elif criteria == "consonant_rich":
        result = []
        for word in words:
            count = 0
            for ch in word.lower():
                if ch.isalpha() and ch not in vowels:
                    count += 1
            if count > 5:
                result.append(word)
        return result

    elif criteria == "sorted":
        result = []
        for word in words:
            if list(word.lower()) == sorted(word.lower()):
                result.append(word)
        return result

    else:
        return None
```
