```python
#adding attributes to the class manually
class Student:
    pass
s1 = Student()
s1.name = "Rashi"
s1.age = 19

s2 = Student()
s2.name ="Ish"
s2.age = 20
print(s1.name,s2.age) 
```

```python
class car:
    pass
c1 = car()
c1.name = 'BMW'
c1.price = "1Cr"

c2 =  car()
c2.name = 'ferrari'
c2.price = '6Cr'

print(c1.name, c2.price)
```

```python
# This is a constructor with a parameter
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

s1 = Student("john",20)
print(s1.name,s1.age)        
```

```python
#object and class with method
class Student:
    def __init__(self,name,age):
        self.name= name
        self.age= age

    def display(self):
        print(f"hello my name is {self.name} and my age is {self.age}")
        print("Nice to meet youu")

s1=Student("john",20)
s1.display()
```
