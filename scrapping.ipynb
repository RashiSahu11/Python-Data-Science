```python
import requests
from bs4 import BeautifulSoup

url ="https://quotes.toscrape.com"
response = requests.get(url) #download website html
print(response.status_code)

#beautifulsoup4 converts html into readable form
soup = BeautifulSoup(response.text,'html.parser')

quotes = soup.find_all("div",class_='quote')

for quote in quotes:
    text = quote.find('span',class_='text').text
    author = quote.find("small",class_='author').text
    print(text)
    print(author)
```

```python
import pandas as pd
url="https://www.flipkart.com/search?q=mobile+phones&otracker=search&otracker1=search&marketplace=FLIPKART&as-show=off&as=off"
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0 Safari/537.36"
    ),
    "Accept-Language": "en-US,en;q=0.9",
}

response = requests.get(url,headers=headers,timeout=20)

print("Status code:" , response.status_code)

if response.status_code == 403:
    print("Website is blocking the request . Please try again later or use a different User-Agent")
else:
    response.raise_for_status()
    soup = BeautifulSoup(response.text,"html.parser")
    print("Page loaded successfully") 


#parent product container
products = soup.find_all("div",class_='lvJbLV')

print("Total products found:",len(products))

#final data list
data=[]


#loop through each product
for product in products:

    #product name
    name=product.find("div",class_='RG5Slk')

    if name:
        name = name.text.strip()
    else:
        name='N/A'

    #price
    price=product.find("div",class_='hZ3P6w DeU9vF')

    if price:
        price = price.text.strip()
    else:    
        price ='N/A'

    #rating
    rating = product.find('div',class_="MKiFS6")

    if rating:
        rating = rating.text.strip()
    else:
        rating='N/A'

    #product link
    link = product.find('div',class_='k7wcnx')     

    if link:
        link = "https://www.flipkart.com"+ link.get("href")
    else:
        link='N/A'

    #store in list
    data.append([name,price,rating,link])

#dataframe
df = pd.DataFrame(
    data,
    columns=['product_name','price','rating','product_link']
)               

#display output
print(df)

#CSV save
df.to_csv("flipkart_product.csv",index=False)

print("Data Saved Successfully")
```
